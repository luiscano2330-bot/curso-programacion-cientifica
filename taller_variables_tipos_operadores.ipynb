# Taller de variables, tipos de datos y operadores

## Ejercicio 1 — Inventario de la tienda

Se declaran las variables del inventario con el tipo de dato correcto: nombre del producto como texto (`str`), precios como entero (`int`) y decimal (`float`), si hay descuentos como booleano (`bool`) y las cantidades como enteros (`int`). Luego se verifica el tipo de cada una con `type()`.

```python
nombre_producto = "Café"
precio_cuaderno = 3500
precio_lapiz = 800
precio_cafe = 6200.50
hay_descuento = False
cantidad_cuadernos = 3
cantidad_lapices = 5

print("Tipo de nombre_producto:", type(nombre_producto))
print("Tipo de precio_cuaderno:", type(precio_cuaderno))
print("Tipo de precio_lapiz:", type(precio_lapiz))
print("Tipo de precio_cafe:", type(precio_cafe))
print("Tipo de hay_descuento:", type(hay_descuento))
print("Tipo de cantidad_cuadernos:", type(cantidad_cuadernos))
print("Tipo de cantidad_lapices:", type(cantidad_lapices))
```

## Ejercicio 2 — Total de la compra y descuento

Se retoman las variables del Ejercicio 1 y se calcula el subtotal de cuadernos, el subtotal de lápices y el total general sumando el precio del café. Sobre el total se aplica un descuento del 10% y se obtiene el total final. Se usan los operadores `*`, `+` y `-`.

```python
total_cuadernos = precio_cuaderno * cantidad_cuadernos
total_lapices = precio_lapiz * cantidad_lapices
total_general = total_cuadernos + total_lapices + precio_cafe
descuento = total_general * 0.10
total_con_descuento = total_general - descuento

print("Total cuadernos:", total_cuadernos)
print("Total lápices:", total_lapices)
print("Total general:", total_general)
print("Descuento del 10%:", descuento)
print("Total con descuento:", total_con_descuento)
```

## Ejercicio 3 — Reparto en paquetes

Con la cantidad de cuadernos se calcula cuántos paquetes completos de 4 cuadernos se pueden armar (división entera `//`) y cuántos cuadernos sueltos quedan (módulo `%`). Además se eleva el precio del café al cuadrado con el operador de potencia `**`.

```python
paquetes_completos = cantidad_cuadernos // 4
cuadernos_sueltos = cantidad_cuadernos % 4
cafe_al_cuadrado = precio_cafe ** 2

print("Paquetes completos de 4 cuadernos:", paquetes_completos)
print("Cuadernos sueltos que quedan:", cuadernos_sueltos)
print("Precio del café elevado al cuadrado:", cafe_al_cuadrado)
```

## Ejercicio 4 — Comparar precios de café

Se comparan los precios de dos marcas de café usando operadores de comparación (`<`, `>` y `==`) para saber cuál es más barata y si cuestan lo mismo. El resultado de cada comparación es un booleano (`True` o `False`).

```python
precio_marca_a = 15900
precio_marca_b = 14500

print("¿La marca A es más barata que la B?", precio_marca_a < precio_marca_b)
print("¿La marca B es más barata que la A?", precio_marca_b < precio_marca_a)
print("¿Ambas marcas cuestan lo mismo?", precio_marca_a == precio_marca_b)
```

## Ejercicio 7 — Factura de servicios y consumo normal

Se calcula el total a pagar de una factura multiplicando el consumo en kWh por la tarifa por kWh. Luego se evalúa por separado si el consumo es mayor o igual a 100 y menor o igual a 200, y se combinan ambas comparaciones con `and` para saber si el consumo es normal.

```python
consumo_kwh = 142
tarifa_kwh = 623.5

total_pagar = consumo_kwh * tarifa_kwh
consumo_mayor_igual_100 = consumo_kwh >= 100
consumo_menor_igual_200 = consumo_kwh <= 200
consumo_normal = consumo_mayor_igual_100 and consumo_menor_igual_200

print("Total a pagar:", total_pagar)
print("¿El consumo es mayor o igual a 100?", consumo_mayor_igual_100)
print("¿El consumo es menor o igual a 200?", consumo_menor_igual_200)
print("¿El consumo es normal (entre 100 y 200)?", consumo_normal)
```

## Ejercicio 8 — Promedio de calificaciones y aprobación

Se calcula el promedio de tres notas parciales con los operadores `+` y `/`, y se compara con la nota de aprobación (3.0) usando el operador `>=`. El resultado de la comparación es un booleano.

```python
nota_parcial_1 = 4.2
nota_parcial_2 = 3.5
nota_parcial_3 = 4.8

promedio_calificaciones = (nota_parcial_1 + nota_parcial_2 + nota_parcial_3) / 3
aprueba = promedio_calificaciones >= 3.0

print("Promedio de calificaciones:", promedio_calificaciones)
print("¿El promedio es mayor o igual a 3.0 (aprueba)?", aprueba)
```
