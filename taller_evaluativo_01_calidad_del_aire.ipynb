## Ejercicio 2 — Conversión de unidades y error frente al equipo patrón

Convierto la lectura del sensor de mg/m3 a ug/m3 multiplicándola por el factor de conversión del ejercicio 1 y calculo el error absoluto, el error relativo porcentual y el error cuadrático frente al valor del equipo patrón.

```python
lectura_sensor_mg = 0.0418
valor_patron_ug = 40.0

lectura_sensor_ug = lectura_sensor_mg * factor_conversion
error_absoluto = lectura_sensor_ug - valor_patron_ug
error_relativo_porcentual = (error_absoluto / valor_patron_ug) * 100
error_cuadratico = error_absoluto ** 2

print("Lectura del sensor en ug/m3:", lectura_sensor_ug)
print("Error absoluto:", error_absoluto)
print("Error relativo porcentual:", error_relativo_porcentual)
print("Error cuadrático:", error_cuadratico)
```

## Ejercicio 6 — Limpiar la serie horaria

Recorro las doce lecturas de la serie. Las que valen -999.0 son fallas del sensor: las cuento y las salto con `continue`. Con las que quedan armo la lista de lecturas válidas, sumo y calculo el promedio al final del recorrido.

```python
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]

lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

for lectura_serie in lecturas_pm25:
    if lectura_serie == -999.0:
        lecturas_descartadas += 1
        continue
    lecturas_validas.append(lectura_serie)
    suma_pm25 += lectura_serie

promedio_pm25 = suma_pm25 / len(lecturas_validas)

print("Lecturas válidas:", len(lecturas_validas))
print("Lecturas descartadas:", lecturas_descartadas)
print("Promedio de PM2.5:", promedio_pm25)
```

## Ejercicio 7 — Estadística descriptiva de la serie limpia

Con las lecturas válidas busco el máximo y el mínimo en un primer recorrido y calculo el rango. En un segundo recorrido sumo las diferencias con el promedio elevadas al cuadrado y saco la desviación estándar con `** 0.5`. El promedio ya debe estar terminado, por eso son dos pasadas.

```python
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

for valor in lecturas_validas:
    if valor > maximo_pm25:
        maximo_pm25 = valor
    if valor < minimo_pm25:
        minimo_pm25 = valor

rango_pm25 = maximo_pm25 - minimo_pm25

suma_desviaciones = 0.0
for valor in lecturas_validas:
    suma_desviaciones += (valor - promedio_pm25) ** 2

desviacion_estandar = (suma_desviaciones / len(lecturas_validas)) ** 0.5

print("Máximo:", maximo_pm25)
print("Mínimo:", minimo_pm25)
print("Rango:", rango_pm25)
print("Desviación estándar:", desviacion_estandar)
```

## Ejercicio 8 — Decaimiento tras cerrar la fuente de emisión

Tras el pico de contaminación la concentración baja 15% cada hora. Uso un `while` para multiplicar la concentración por el factor de decaimiento hasta que quede en categoría Buena (12.0 o menos), contando las horas que tarda.

```python
concentracion = 58.3
factor_decaimiento = 0.85
umbral_buena = 12.0

horas_transcurridas = 0

while concentracion > umbral_buena:
    concentracion *= factor_decaimiento
    horas_transcurridas += 1

print("Horas transcurridas:", horas_transcurridas)
print("Concentración final:", concentracion)
```

## Ejercicio 9 — Registros de varias estaciones

Recorro la lista de estaciones, cada una representada como un diccionario. Desempaqueto sus coordenadas en latitud y longitud, calculo el promedio de sus lecturas con un `for` interno, clasifico ese promedio con la misma escala y traigo la humedad con `get` para que las estaciones sin ese sensor no fallen.

```python
registros_estaciones = [
    {"codigo": "EST-01", "sector": "Norte",  "coordenadas": (6.28, -75.57), "lecturas": [8.4, 12.7, 15.2]},
    {"codigo": "EST-02", "sector": "Centro", "coordenadas": (6.24, -75.58), "lecturas": [41.8, 58.3, 36.1]},
    {"codigo": "EST-03", "sector": "Sur",    "coordenadas": (6.19, -75.59), "lecturas": [9.6, 13.9, 7.5], "humedad_relativa": 68.0},
]

for estacion in registros_estaciones:
    latitud, longitud = estacion["coordenadas"]

    suma_lecturas = 0.0
    for lectura_estacion in estacion["lecturas"]:
        suma_lecturas += lectura_estacion
    promedio_estacion = suma_lecturas / len(estacion["lecturas"])

    humedad_estacion = estacion.get("humedad_relativa", "no disponible")

    if promedio_estacion <= 12.0:
        categoria_estacion = "Buena"
    elif promedio_estacion <= 35.4:
        categoria_estacion = "Moderada"
    elif promedio_estacion <= 55.4:
        categoria_estacion = "Dañina para grupos sensibles"
    else:
        categoria_estacion = "Dañina para la salud"

    print(f"Estación {estacion['codigo']} ({estacion['sector']}) — Lat: {latitud}, Lon: {longitud} — Promedio: {promedio_estacion} — Categoría: {categoria_estacion} — Humedad: {humedad_estacion}")
```
