# Taller evaluativo 03 — NumPy


## Ejercicio 1 — Crear los arreglos y leer su etiqueta

Convierto la tabla de lecturas en un arreglo 2D con `np.array` y leo sus atributos `shape`, `ndim`, `size` y `dtype`. También creo los días con `np.arange` y los arreglos de pendientes e interceptos de cada sensor.

```python
import numpy as np

lecturas_crudas = np.array([
    [10.0, 14.5, 22.0, 30.5, 28.0, 18.0, 9.5],
    [12.0, 20.0, 35.0, 48.0, 41.0, 25.0, 14.0],
    [8.0, 11.0, 19.0, 27.0, 33.0, 21.0, 6.0],
    [15.0, 24.0, 31.0, 44.0, 52.0, 36.0, 17.0],
    [5.0, 9.0, 16.0, 22.0, 45.0, 30.0, 4.0],
])

forma = lecturas_crudas.shape
dimensiones = lecturas_crudas.ndim
total = lecturas_crudas.size
tipo = lecturas_crudas.dtype

dias = np.arange(1, 8)
pendientes = np.array([1.2, 0.9, 1.1, 1.0, 1.3])
interceptos = np.array([-0.5, 1.0, -2.0, 0.5, -1.5])

print("Forma:", forma)
print("Dimensiones:", dimensiones)
print("Total de elementos:", total)
print("Tipo de dato:", tipo)
print("Días:", dias)
print("Pendientes:", pendientes)
print("Interceptos:", interceptos)
```

## Ejercicio 2 — Señalar celdas y recortar bloques

Leo partes de la tabla con indexación y slicing: una celda suelta, el último sensor con índice negativo, una columna completa (un día), una fila completa (un sensor) y un bloque rectangular. Python cuenta desde 0 y el fin de un slice no se incluye.

```python
lectura_c_dia4 = lecturas_crudas[2, 3]
ultimo_primer = lecturas_crudas[-1, 0]
dia5 = lecturas_crudas[:, 4]
semana_b = lecturas_crudas[1, :]
bloque = lecturas_crudas[1:4, 2:6]

print("SBC-C en el día 4:", lectura_c_dia4)
print("Último sensor en el día 1:", ultimo_primer)
print("Día 5 (columna completa):", dia5)
print("Semana de SBC-B (fila completa):", semana_b)
print("Bloque SBC-B a SBC-D, días 3 a 6:")
print(bloque)
print("Forma del bloque:", bloque.shape)
```

## Ejercicio 3 — Limpiar una campaña con máscaras booleanas

Creo la serie de campaña como arreglo y dos máscaras: `faltantes`, que marca el centinela -999.0, y `es_util`, que exige estar entre 0 y 500.0 (las dos comparaciones van entre paréntesis y unidas con `&`). Filtro con la máscara como índice y calculo el promedio de las lecturas útiles.

```python
campana_sbc_a = np.array([10.0, -999.0, 25.0, 40.0, -999.0, 8.0, 52.0, 600.0])

faltantes = campana_sbc_a == -999.0
es_util = (campana_sbc_a >= 0) & (campana_sbc_a <= 500.0)

lecturas_utiles = campana_sbc_a[es_util]
promedio_utiles = lecturas_utiles.mean()

print("Lecturas faltantes:", int(faltantes.sum()))
print("Lecturas útiles:", lecturas_utiles)
print("Promedio de las lecturas útiles:", promedio_utiles)
```

## Ejercicio 4 — Calibrar todo con operaciones vectorizadas y broadcasting

Calibro las 35 lecturas de una sola vez. La línea `lecturas_crudas * pendientes` falla con `ValueError` porque las formas son (5, 7) y (5,): NumPy alinea las dimensiones de derecha a izquierda, compara el 7 con el 5 y no coinciden. Por eso convierto `pendientes` e `interceptos` en columnas de forma (5, 1) con `reshape`, y así el 1 se estira sobre los 7 días (broadcasting).

```python
pendientes_col = pendientes.reshape(5, 1)
interceptos_col = interceptos.reshape(5, 1)

calibradas = lecturas_crudas * pendientes_col + interceptos_col

campana_calibrada = lecturas_utiles * 1.2 - 0.5
promedio_calibrado_sbc_a = campana_calibrada.mean()

print("Forma de calibradas:", calibradas.shape)
print("Fila de SBC-A:", calibradas[0])
print("Campaña calibrada de SBC-A:", campana_calibrada)
print("Promedio calibrado de SBC-A:", promedio_calibrado_sbc_a)
```

## Ejercicio 5 — Resumir la semana con estadística por eje

Resumo la red completa y por dirección: `axis=1` da un valor por sensor (5) y `axis=0` un valor por día (7). También obtengo el máximo por sensor y el promedio de las lecturas que superan 35.4 usando una máscara. Todo con métodos de NumPy, sin sumar a mano.

```python
promedio_general = calibradas.mean()
promedio_por_sensor = calibradas.mean(axis=1)
promedio_por_dia = calibradas.mean(axis=0)
maximo_por_sensor = calibradas.max(axis=1)
media_sobre_limite = calibradas[calibradas > 35.4].mean()

print("Promedio general:", np.round(promedio_general, 2))
print("Promedio por sensor:", np.round(promedio_por_sensor, 2))
print("Promedio por día:", np.round(promedio_por_dia, 2))
print("Máximo por sensor:", np.round(maximo_por_sensor, 2))
print("Media de las lecturas sobre 35.4:", np.round(media_sobre_limite, 2))
```

## Ejercicio 6 — Funciones universales y reemplazo con máscara

Mido cuánto se pasó cada sensor del límite de 35.4 con `np.maximum`, calculo la distancia absoluta de SBC-B con `np.abs` y redondeo con `np.round`. Para `en_alerta` copio `calibradas` con `.copy()` y reemplazo por 0.0 lo que no supera el límite, así el arreglo original queda intacto.

```python
excedente = np.maximum(calibradas - 35.4, 0)
excedente_por_sensor = np.round(excedente.sum(axis=1), 1)
desvio_sensor_b = np.round(np.abs(calibradas[1] - 35.4), 1)

en_alerta = calibradas.copy()
en_alerta[en_alerta <= 35.4] = 0.0

print("Excedente por sensor:", excedente_por_sensor)
print("Desvío de SBC-B frente a 35.4:", desvio_sensor_b)
print("Celdas en alerta:", int((en_alerta > 0).sum()))
```

## Ejercicio 7 — Integración: el informe de la red

Cuento con máscaras cuántas de las 35 lecturas calibradas caen en cada categoría de la escala, cuidando que los intervalos no se solapen, y busco el día más contaminado filtrando `dias` con una máscara sobre `promedio_por_dia`.

```python
buenas = int((calibradas <= 12.0).sum())
moderadas = int(((calibradas > 12.0) & (calibradas <= 35.4)).sum())
daninas_sensibles = int(((calibradas > 35.4) & (calibradas <= 55.4)).sum())
daninas_salud = int((calibradas > 55.4).sum())

dia_pico = dias[promedio_por_dia == promedio_por_dia.max()][0]

print(f"Lecturas por categoría: Buena={buenas}, Moderada={moderadas}, Dañina para grupos sensibles={daninas_sensibles}, Dañina para la salud={daninas_salud}")
print(f"Día más contaminado: {dia_pico} con {round(promedio_por_dia.max(), 1)} ug/m3")
```

## Reto de extensión (opcional)

Compruebo los mismos cuatro conteos con `np.digitize`, que asigna a cada lectura un índice de categoría de 0 a 3.

```python
indices_categoria = np.digitize(calibradas, [12.0, 35.4, 55.4], right=True)

print("Con digitize -> Buena:", int((indices_categoria == 0).sum()))
print("Con digitize -> Moderada:", int((indices_categoria == 1).sum()))
print("Con digitize -> Dañina para grupos sensibles:", int((indices_categoria == 2).sum()))
print("Con digitize -> Dañina para la salud:", int((indices_categoria == 3).sum()))
```
